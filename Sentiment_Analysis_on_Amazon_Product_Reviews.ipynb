```python
import kagglehub
import os
import pandas as pd

# download latest version
path = kagglehub.dataset_download("datafiniti/consumer-reviews-of-amazon-products")
print("Path to dataset files:", path)

# list files in the downloaded path
print("Files in the dataset directory:")
csv_files = []
for root, dirs, files in os.walk(path):
    for file in files:
        print(os.path.join(root, file))
        if file.endswith('.csv'):
            csv_files.append(os.path.join(root, file))

# assuming there's a CSV file, let's try to load one
if csv_files:
    # prioritize 'Datafiniti_Amazon_Consumer_Reviews_of_Amazon_Products.csv' if available
    target_csv_file = None
    for f in csv_files:
        if 'Datafiniti_Amazon_Consumer_Reviews_of_Amazon_Products.csv' in f:
            target_csv_file = f
            break
    if not target_csv_file:
        target_csv_file = csv_files[0] # Fallback to the first CSV found

    print(f"\nLoading CSV file: {target_csv_file}")
    try:
        df = pd.read_csv(target_csv_file)
        print("Successfully loaded CSV file into a DataFrame.")
        print("First 5 rows of the DataFrame:")
        display(df.head())
    except Exception as e:
        print(f"Error loading CSV file: {e}")
else:
    print("No CSV files found in the downloaded dataset.")
```

```python
# import necessary libraries

import pandas as pd
import numpy as np
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

```

```python
# download necessary NLTK data

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
```

```python
# preprocess the data

# fill any potential NaN values in 'reviews.text' with empty strings
df['reviews.text'] = df['reviews.text'].fillna('')

# convert text to lowercase
df['processed_reviews'] = df['reviews.text'].apply(lambda x: x.lower())

# tokenize text and remove stopwords
stop_words = set(stopwords.words('english'))
df['processed_reviews'] = df['processed_reviews'].apply(lambda x: [
    word for word in word_tokenize(x) if word.isalpha() and word not in stop_words
])

# join tokenized words back into a string
df['processed_reviews'] = df['processed_reviews'].apply(lambda x: ' '.join(x))
```

```python
# split the data into training and testing sets

# using the processed reviews as features (X) and 'reviews.rating' as target (y)
X_train, X_test, y_train, y_test = train_test_split(
    df['processed_reviews'],
    df['reviews.rating'].apply(lambda x: 1 if x > 4 else 0), # Binarize ratings: 1 for >4 stars, 0 otherwise
    test_size=0.2,
    random_state=42
)
```

```python
# create a TF_IDF vectorizer

vectorizer = TfidfVectorizer()
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)
```

```python
# train a Naive Bayes model

# initialize the Multinomial Naive Bayes classifier (corrected typo)
clf = MultinomialNB()

# train the classifier using the TF-IDF transformed training data (corrected typo)
clf.fit(X_train_tfidf, y_train)
```

```python
# evaluate the model's performance

y_pred = clf.predict(X_test_tfidf)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Classification Report:")
print(classification_report(y_test, y_pred))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```
